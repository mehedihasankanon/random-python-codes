# Tensors

Any arrays can be called tensors (they have a more sophisticated defn but let's go with this for now).

- Tensor Dimension: directions the tensor spans.
- 0D tensor: scalar (single number)
- 1D tensor: vector (list of numbers)
- 2D tensor: matrix (table of numbers)
- 3D tensor: cube of numbers (e.g. RGB image)
- nD tensor: tensor with n dimensions (e.g. video with frames, height, width, color channels)

## Why Tensors?
- efficient mathematical operations
- efficient gpu computations, optimized for HW accelerations and parallelization
- real world data is often represented as tensors




# Creating a Tensor

```python
# using empty

a = torch.empty(2, 3) # makes a 2x3 matrix. allocates memory but doesn't allcate any values. quite like C arrays!

type(a)
```

```python
# create a tensor of zeros and ones

b = torch.zeros(2, 3)
c = torch.ones(2, 3)

print(b, c)
```

```python
# random tensors

d = torch.rand(2, 3)

print(d)

# to always generate a same random tensor, we can set the random seed

torch.manual_seed(330)
e = torch.rand(2, 3)

print(e)
```

```python
# custom tensor

f = torch.tensor([[1, 2, 3], [4, 5, 6]])

# you can pass any form of python iterable to the tensor constructor

g = torch.arange(0, 10, 2) # start, end, step

print(g)

h = torch.linspace(0, 10, steps=10) # start, end, number of steps

print(h)

i = torch.eye(5) # identity matrix of size 5x5

print(i)

j = torch.full((2, 3), 7) # creates a 2x3 matrix filled with the value 7

print(j)

k = torch.empty_like(j) # creates an empty tensor with the same shape and dtype as j
# remember that this doesn't allocate any values.
print(k)


# torch.rand_like() doesn't work!!

# we have to specify dtype 

```

**Important: look at the data type of the tensor. By default, PyTorch creates tensors with `float32` data type. You can specify the data type when creating a tensor using the `dtype` argument. For example, to create a tensor of integers, you can use `dtype=torch.int32`.**

Moreover, we can use `torch.to()` to change the data type of an existing tensor. For example, if you have a tensor `a` and you want to convert it to integers, you can do `a = a.to(torch.int32)`.

```python
l = torch.rand_like(j, dtype=torch.float32)

print(l)
```

# Operations on Tensors

```python
ten = torch.rand(2, 3)

print("ten:", ten)
```

```python
# integer ops

# addition

print(f"ten + 2: {ten + 2}")

# deletion

print(f"ten - 2: {ten - 2}")

# multiplication

print(f"ten * 2: {ten * 2}")

# division

print(f"ten / 2: {ten / 2}")

# mod

print(f"ten * 20: {ten * 20}")
print(f"(ten * 20) % 2: {(ten * 20) % 2}")

# abs

print(f"abs(ten * 20): {torch.abs(ten- 2)}")

```

```python
# tensor ops

torch.manual_seed(10101)
a1 = torch.rand(2, 3) * 100
a2 = torch.rand(2, 3) * 100

print(a1)
print(a2)

# element-wise addition

print(f"\na1 + a2: {a1 + a2}")

# element-wise subtraction

print(f"\na1 - a2: {a1 - a2}")

# element-wise multiplication

print(f"\na1 * a2: {a1 * a2}")

# element-wise division [integer division also possible via ``//``]

print(f"\na1 / a2: {a1 / a2}")

# floor, ceil, round also available

# clamping

print(f"\na1 clamped between 20 and 80: {torch.clamp(a1, min=20, max=80)}")
```

```python
# stuff inside torch entries

b1 = torch.randint(size=(2,3),  low=80, high=100)

print(b1)


print("tensor sum:", torch.sum(b1))
print(f"column sum: {torch.sum(b1, dim=0)}")
print(f"row sum: {torch.sum(b1, dim=1)}")

print("")

# torch.mean() --> must be float tensor
# we can pass `dim=` to specify the dimension along which to compute the mean too!

print(f"tensor max: {torch.max(b1)}")
print(f"column max: {torch.max(b1, dim=0)}")
print(f"row max: {torch.max(b1, dim=1)}")

# notice how the row and column max functions return the indices where the max value was found as well.
# useful thing!

# for global location of max value, we can use torch.argmax()
print(f"\nglobal max value: {torch.max(b1)}")
print(f"global max index: {torch.argmax(b1)}")

# for min, we use torch.min() and torch.argmin()
print(f"\nglobal min value: {torch.min(b1)}")
print(f"global min index: {torch.argmin(b1)}")
```

## Matrix operations for tensors

- `torch.matmul()` is used for matrix multiplication, while `torch.mm()` is used for matrix multiplication of 2D tensors. The main difference is that `torch.matmul()` can handle tensors of higher dimensions (e.g., 3D or more) and will perform the appropriate matrix multiplication based on the dimensions of the input tensors, while `torch.mm()` is specifically designed for 2D tensors and will raise an error if the input tensors are not 2D. Meanwhile `torch.bmm()` is used for batch matrix multiplication, which allows you to perform matrix multiplication on batches of matrices. It takes two 3D tensors as input, where the first dimension represents the batch size, and performs matrix multiplication on each pair of matrices in the batch.

- `torch.transspose()`, `torch.det()`, `torch.inverse()`, `torch.eig()`, `torch.svd()`, `torch.qr()` available!


- `torch.dot()` is used for dot product of 1D tensors (vectors), while `torch.cross()` is used for cross product of 3D vectors. The main difference is that `torch.dot()` computes the sum of the products of the corresponding elements of two vectors, resulting in a scalar value, while `torch.cross()` computes the cross product of two 3D vectors, resulting in a new vector that is perpendicular to both input vectors.

    Note that we must specify the dimension along which to perform the cross product when using `torch.cross()`, as it can operate on tensors of higher dimensions. For example, if we have two 2D tensors `a1` and `a2`, we can compute their cross product along the second dimension (rows) using `torch.cross(a1, a2, dim=1)`.

```python
# raises an error since we CANNOT dot product 2D tensors using torch.dot()
# print(f"a1 dot a2: {torch.dot(a1, a2)}")

# flatten() -> flattens the tensor to 1D

# so, tensor([[98.6526, 59.7209, 87.7269],
#            [31.8950, 57.4184, 77.8326]])
# is reduced to tensor([98.6526, 59.7209, 87.7269, 31.8950, 57.4184, 77.8326])

print(f"a1 dot a2: {torch.dot(a1.flatten(), a2.flatten())}")

# `reshape()` can also be used to flatten the tensor
# but it creates a new tensor. 
```

```python
# raises an error since we cannot cross 2D vectors using torch.cross()

# print("a1 cross a2: ", torch.cross(a1, a2, dim=0)) # cross product along the first dimension (columns)

# this is perfect

print("a1 cross a2: ", torch.cross(a1, a2, dim=1)) # cross product along the second dimension (rows)
```

### Comparison ops 

- when we do a comparison op, we get a tensor of boolean values (True/False) in the o/p tensor

- `<`, `>`, `==`, `!=`, `<=`, `>=` etc are available as operators. functions: `torch.eq()`, `torch.ne()`, `torch.gt()`, `torch.lt()`, `torch.ge()`, `torch.le()`.

### Other mathematical operations

- `torch.log()`, `torch.exp()`, `torch.sqrt()`, `torch.sin()`, `torch.cos()`, `torch.tan()` etc.

# Inplace ops

- when we do a entry by entry op, we always get a new tensor as o/p. this will hog a lot of memory for super large tensors

- say we have two tensors `a` and `b`, and we want to add them together. If we do `c = a + b`, we get a new tensor `c` that contains the result of the addition. However, if we want to perform the addition in place (i.e., modify `a` directly), we can use the `add_()` method: `a.add_(b)`. This will add `b` to `a` and store the result back in `a`, without creating a new tensor.

- other functions: `sub_()`, `mul_()`, `div_()`, `pow_()`, `sqrt_()`, `log_()`, `exp_()`, `relu_()` etc.

- basically underscore at the end of operation name, and make that an inplace operation! :3

### Copying tensors

- simply doing `a = b` just assigns the memory address of `b` to `a`, so when we modify `b`, `a` is also modified.
- to create an actual copy of the tensor, we can use the `clone()` method: `a = b.clone()`. This creates a new tensor with the same values as `b`, but with a different memory address.
