# Tensors

Any arrays can be called tensors (they have a more sophisticated defn but let's go with this for now).

- Tensor Dimension: directions the tensor spans.
- 0D tensor: scalar (single number)
- 1D tensor: vector (list of numbers)
- 2D tensor: matrix (table of numbers)
- 3D tensor: cube of numbers (e.g. RGB image)
- nD tensor: tensor with n dimensions (e.g. video with frames, height, width, color channels)

## Why Tensors?
- efficient mathematical operations
- efficient gpu computations, optimized for HW accelerations and parallelization
- real world data is often represented as tensors




# Creating a Tensor

```python
# using empty

a = torch.empty(2, 3) # makes a 2x3 matrix. allocates memory but doesn't allcate any values. quite like C arrays!

type(a)
```

```python
# create a tensor of zeros and ones

b = torch.zeros(2, 3)
c = torch.ones(2, 3)

print(b, c)
```

```python
# random tensors

d = torch.rand(2, 3)

print(d)

# to always generate a same random tensor, we can set the random seed

torch.manual_seed(330)
e = torch.rand(2, 3)

print(e)
```

```python
# custom tensor

f = torch.tensor([[1, 2, 3], [4, 5, 6]])

# you can pass any form of python iterable to the tensor constructor

g = torch.arange(0, 10, 2) # start, end, step

print(g)

h = torch.linspace(0, 10, steps=10) # start, end, number of steps

print(h)

i = torch.eye(5) # identity matrix of size 5x5

print(i)

j = torch.full((2, 3), 7) # creates a 2x3 matrix filled with the value 7

print(j)

k = torch.empty_like(j) # creates an empty tensor with the same shape and dtype as j
# remember that this doesn't allocate any values.
print(k)


# torch.rand_like() doesn't work!!

# we have to specify dtype 

```

**Important: look at the data type of the tensor. By default, PyTorch creates tensors with `float32` data type. You can specify the data type when creating a tensor using the `dtype` argument. For example, to create a tensor of integers, you can use `dtype=torch.int32`.**

Moreover, we can use `torch.to()` to change the data type of an existing tensor. For example, if you have a tensor `a` and you want to convert it to integers, you can do `a = a.to(torch.int32)`.
